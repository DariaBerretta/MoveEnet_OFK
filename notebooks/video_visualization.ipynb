```python
import cv2
import numpy as np
import torch
import os
from pathlib import Path
from torch_geometric.data import Batch
from graph_enet.pyScarf.scarf.scarf_class import SCARF
from graph_enet.utils.log_loader import load_events_from_log, load_skeleton_from_log
from graph_enet.data.graph_builder_splineConv import build_scarf_graph_splineConv
from graph_enet.hpe_gnn.model.hpegnn import hpeGnn_splineConv_single_weight
from IPython.display import Video
import matplotlib.pyplot as plt

# Set up plotting style
plt.style.use("seaborn-v0_8")
```

```python
# === Configuration ===
# Run Label (folder name in lightning_logs)
RUN_LABEL = "scarf_dataset_test20251210" 

# Data Path (Path to the raw clip folder containing ch0dvs and ch0GT50Hzskeleton)
DATA_PATH = "/home/dberretta-iit.local/data/new_scarfGNN_val/raw/cam2_S1_Directions"

# Video Settings
VIDEO_OUTPUT = "prediction_video2.mp4"
DURATION = 5.0 # seconds to process
START_TIME = 0.0 # start time in seconds

# Skeleton Connections (from training_metrics_dashboard.ipynb)
# These indices are used directly to index into the joint array
skeleton_connections = [
    (1, 2), (1, 3), (1, 6), (2, 4), (2, 5), (3, 7), (4, 8), (5, 6), (5, 10), (6, 9), (9, 11), (10, 12)
]
```

```python
# === Setup: Find Checkpoint ===
def find_project_root(start: Path, marker: str = "lightning_logs") -> Path:
    """Walk upward from start until a folder containing `marker` is found."""
    for p in [start] + list(start.parents):
        if (p / marker).exists():
            return p
    raise FileNotFoundError(f"Could not find `{marker}` above {start}")

# Locate project root and run directory
project_root = find_project_root(Path.cwd())
log_root = project_root / "lightning_logs"
run_dir = log_root / RUN_LABEL

if not run_dir.exists():
    raise FileNotFoundError(f"Run folder not found: {run_dir}")

# Find latest version
versions = sorted([p for p in run_dir.glob('version_*') if p.is_dir()], key=lambda p: p.stat().st_mtime)
if not versions:
    raise RuntimeError(f"No version_* folders inside {run_dir}")
latest_version = versions[-1]
print(f"Using run version: {latest_version.name}")

# Find latest checkpoint
ckpt_dir = latest_version / 'checkpoints'
if not ckpt_dir.exists():
     raise RuntimeError(f"Checkpoints folder not found: {ckpt_dir}")
     
checkpoints = sorted(ckpt_dir.glob('*.ckpt'), key=lambda p: p.stat().st_mtime)
if not checkpoints:
    raise RuntimeError(f"No checkpoints found in {ckpt_dir}")
latest_ckpt = checkpoints[-1]
print(f"Using checkpoint: {latest_ckpt}")
```

```python
# === Helper Functions ===
def find_closest_gt(timestamp, gt_data):
    """Find the closest ground truth skeleton for a given timestamp."""
    idx = np.searchsorted(gt_data['ts'], timestamp)
    if idx == 0: return gt_data['keypoints'][0]
    if idx >= len(gt_data): return gt_data['keypoints'][-1]
    
    prev_ts = gt_data['ts'][idx-1]
    next_ts = gt_data['ts'][idx]
    
    if abs(timestamp - prev_ts) < abs(timestamp - next_ts):
        return gt_data['keypoints'][idx-1]
    else:
        return gt_data['keypoints'][idx]

def draw_skeleton_cv2(img, joints, connections, color, thickness=2):
    """Draw skeleton on an OpenCV image."""
    joints = joints.reshape(-1, 2)
    
    # Draw connections
    for u, v in connections:
        if u < len(joints) and v < len(joints):
            pt1 = (int(joints[u][0]), int(joints[u][1]))
            pt2 = (int(joints[v][0]), int(joints[v][1]))
            cv2.line(img, pt1, pt2, color, thickness)
            
    # Draw joints
    for i in range(len(joints)):
        cv2.circle(img, (int(joints[i][0]), int(joints[i][1])), 3, color, -1)
```

# === Helper Functions ===
Run this cell to define the necessary helper functions before running the main processing loop.

```python
# === Main Processing ===

# 1. Load Model
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Running on device: {device}")

print("Loading model...")
model = hpeGnn_splineConv_single_weight.load_from_checkpoint(str(latest_ckpt), map_location=device)
model.to(device)
model.eval()

# 2. Load Data
print(f"Loading data from {DATA_PATH}...")
events = load_events_from_log(os.path.join(DATA_PATH, 'ch0dvs'))
gt_skel = load_skeleton_from_log(os.path.join(DATA_PATH, 'ch0GT50Hzskeleton'))
print(f"Loaded {len(events)} events and {len(gt_skel)} skeleton frames.")

# 3. Initialize SCARF
rf_size = 14
alpha = 1.0
C = 0.3
res = (640, 480)
dt = 0.01
scarf = SCARF(res, rf_size, alpha, C)

# 4. Initialize Video Writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
fps = int(1.0 / dt)
out = cv2.VideoWriter(VIDEO_OUTPUT, fourcc, fps, res)

# 5. Processing Loop
start_ts = events['ts'][0] + START_TIME
end_ts = min(events['ts'][-1], start_ts + DURATION)
timer = start_ts
idx = np.searchsorted(events['ts'], start_ts)
N = len(events)
frame_count = 0

print(f"Generating video ({DURATION}s)...")
try:
    while timer < end_ts:
        # Update SCARF with events in this time window
        start_idx = idx
        while idx < N and events['ts'][idx] <= timer:
            idx += 1
        
        # Process batch of events
        batch_events = events[start_idx:idx]
        for ev in batch_events:
            scarf.update(ev['x'], ev['y'], ev['pol'])
            
        # Get SCARF Image
        img32 = scarf.get_surface()
        img8U = (img32 * 255).clip(0, 255).astype('uint8')
        inverted = cv2.bitwise_not(img8U)
        frame = cv2.cvtColor(inverted, cv2.COLOR_GRAY2BGR)
        
        # Get Ground Truth
        gt_joints = find_closest_gt(timer, gt_skel)
        
        # Build Graph & Predict
        # Note: We pass gt_joints to build_scarf_graph_splineConv because it requires a skeleton 
        # to construct the 'y' target, even though we are doing inference.
        graph_data = build_scarf_graph_splineConv(scarf, gt_joints, device=device)
        
        if graph_data is not None:
            # Prepare batch for model
            graph_data.batch = torch.zeros(graph_data.num_nodes, dtype=torch.long, device=device)
            batch = Batch.from_data_list([graph_data]).to(device)
            
            # Run Inference
            with torch.no_grad():
                pred, _ = model(batch.x, batch.edge_index, batch.edge_attr, batch.pos, batch.batch)
                pred_joints = pred.cpu().numpy().flatten()
            
            # Draw Skeletons
            draw_skeleton_cv2(frame, gt_joints, skeleton_connections, (0, 255, 0)) # Green: Ground Truth
            draw_skeleton_cv2(frame, pred_joints, skeleton_connections, (0, 0, 255)) # Red: Prediction
            
        # Add timestamp text
        cv2.putText(frame, f"T: {timer:.2f}s", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2)
        
        # Write frame
        out.write(frame)
        
        timer += dt
        frame_count += 1
        
        if frame_count % 50 == 0:
            print(f".", end="", flush=True)

finally:
    out.release()
    print(f"\nVideo saved to {VIDEO_OUTPUT}")
```
